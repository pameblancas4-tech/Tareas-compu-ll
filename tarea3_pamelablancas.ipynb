```python
import pandas as pd 

orders = pd.read_csv("olist_orders_dataset.csv")
order_items = pd.read_csv("olist_order_items_dataset.csv")
order_reviews = pd.read_csv("olist_order_reviews_dataset.csv")
products = pd.read_csv("olist_products_dataset.csv")
customers = pd.read_csv("olist_customers_dataset.csv")
sellers = pd.read_csv("olist_sellers_dataset.csv")
geo = pd.read_csv("olist_geolocation_dataset.csv")
payments = pd.read_csv("olist_order_payments_dataset.csv")
```

```python
#conversión de las columnas de las fechas de orders

orders["order_purchase_timestamp"] = pd.to_datetime(
    orders["order_purchase_timestamp"],
    format="%Y-%m-%d %H:%M:%S",
    errors="coerce"
)

orders["order_delivered_customer_date"] = pd.to_datetime(
    orders["order_delivered_customer_date"],
    format="%Y-%m-%d %H:%M:%S",
    errors="coerce"
)

orders["order_estimated_delivery_date"] = pd.to_datetime(
    orders["order_estimated_delivery_date"],
    format="%Y-%m-%d %H:%M:%S",
    errors="coerce"
)
```

### Ejercicio 16 

```python
order_items["costo_total"] = order_items.apply(lambda fila: (fila["price"]+ fila["freight_value"]) , axis = 1)
order_items["total_caro_barato"] =order_items["costo_total"].apply(lambda x: "Caro" if x > 100 else "Barato")
df_16 = order_items.merge(products, on = "product_id", how = "left")
df_16.pivot_table(columns = "total_caro_barato", index = "product_category_name", values = "costo_total", aggfunc= "mean")
```

### Ejercicio 17

```python
order_items["envio_respecto_precio"] = order_items.apply(lambda fila: fila["freight_value"]/fila["price"], axis = 1)
order_items["Clasificacion"] = order_items["envio_respecto_precio"].apply(lambda x: "envio alto" if x > 0.20 else "envio bajo")
df_17 = order_items.merge(products, on = "product_id", how = "left")
df_17.pivot_table(index = "product_category_name", values = "freight_value", columns = "Clasificacion", aggfunc = "mean")
```

### Ejercicio 18

```python
promedio_global = order_items["price"].mean()
order_items["precio_superior_promedio"] = order_items["price"].apply(lambda x: "Superior" if x > promedio_global else "Inferior")
df_18 = order_items.merge(products, on="product_id", how="left")
df_18.pivot_table(index="product_category_name", columns="precio_superior_promedio", values="price", aggfunc="count")
```

### Ejercicio 20

```python
#¿Qué categorías concentran más artículos con envío alto (más del 20% del precio)?
order_items["envio_respecto_precio"] = order_items.apply(lambda fila: fila["freight_value"]/fila["price"], axis = 1)
order_items["Clasificacion"] = order_items["envio_respecto_precio"].apply(lambda x: "envio alto" if x > 0.20 else "envio bajo")
#Aunque esta ya estaba hecha en un problema anterior, la adjunto para cumplir con la rubrica
df_20 = order_items.merge(products, on="product_id", how="left")
df_20.pivot_table(values = "order_id", index = "product_category_name",columns = "Clasificacion", aggfunc = "count").sort_values("envio alto", ascending = False).head(10)
#Las categorías con más artículos de envío alto son cama_mesa_banho, moveis_decoracao y beleza_saude, pero eso se debe en parte a su tamaño. Al comparar contra los de envío bajo, telefonia y eletronicos destacan
```

```python
df_22 = df_20.merge(orders[["order_id", "order_status"]], on="order_id", how="left")
df_22 = df_22[df_22["order_status"].isin(["delivered", "canceled"])]
df_22["envio_pct"] = df_22["envio_respecto_precio"].apply(lambda x: x * 100)

tabla22a = df_22.pivot_table(values="price", index="product_category_name", columns="order_status", aggfunc="mean")
tabla22b = df_22.pivot_table(values="envio_pct", index="product_category_name", columns="order_status", aggfunc="mean")
display(tabla22a, tabla22b)
```
